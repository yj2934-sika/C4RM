## 2. getBondPrice

The bond price equals the sum of the present values of all
coupon payments and the principal repayment.

The function adjusts the number of periods, coupon payment,
and discount rate for the payment frequency.

```python
def getBondPrice(y, face, couponRate, m, ppy=1):
    # Payment periods: 1, 2, ..., m * ppy
    periods = np.arange(1, int(m * ppy) + 1)

    # Coupon payment in each period
    cash_flows = np.full(
        periods.size,
        face * couponRate / ppy,
        dtype=float
    )

    # Repay principal in the final period
    cash_flows[-1] += face

    # Discount all cash flows and add them
    present_values = cash_flows / (1 + y / ppy) ** periods

    return np.sum(present_values)
```

```python
y = 0.03
face = 2000000
couponRate = 0.04
m = 10

annual_price = getBondPrice(y, face, couponRate, m)
semiannual_price = getBondPrice(y, face, couponRate, m, ppy=2)

print(f"Annual-payment bond price: {annual_price:,.6f}")
print(f"Semiannual-payment bond price: {semiannual_price:,.6f}")
```

## 3. getBondDuration

Macaulay duration is the present-value-weighted average time
to receive the bond's cash flows.

Payment times are measured in years, so the result is in years.

```python
def getBondDuration(y, face, couponRate, m, ppy=1):
    # Payment periods and payment times in years
    periods = np.arange(1, int(m * ppy) + 1)
    times = periods / ppy

    # Coupon payments and final principal repayment
    cash_flows = np.full(
        periods.size,
        face * couponRate / ppy,
        dtype=float
    )
    cash_flows[-1] += face

    # Present values of all cash flows
    present_values = cash_flows / (1 + y / ppy) ** periods

    # Present-value-weighted average payment time
    duration = np.sum(times * present_values) / np.sum(present_values)

    return duration
```

```python
annual_duration = getBondDuration(
    y=0.03,
    face=2000000,
    couponRate=0.04,
    m=10
)

semiannual_duration = getBondDuration(
    y=0.03,
    face=2000000,
    couponRate=0.04,
    m=10,
    ppy=2
)

print(f"Annual-payment Macaulay duration: {annual_duration:.6f} years")
print(f"Semiannual-payment Macaulay duration: {semiannual_duration:.6f} years")
```

## 4. FizzBuzz

The function processes integers from start up to, but not including, finish.

Multiples of 3 become "fizz", multiples of 5 become "buzz", and multiples of both become "fizzbuzz". Other values remain integers.

An object array stores both integers and strings.

```python
def FizzBuzz(start, finish):
    # Generate integers; finish is excluded
    numbers = np.arange(start, finish)

    # Store both integers and strings
    result = np.array(numbers, dtype=object)

    # Replace multiples using Boolean masks
    result[numbers % 3 == 0] = "fizz"
    result[numbers % 5 == 0] = "buzz"
    result[numbers % 15 == 0] = "fizzbuzz"

    return result
```

```python
print("FizzBuzz from 1 to 15:")
print(FizzBuzz(1, 16))

print("\nFizzBuzz from 16 to 30:")
print(FizzBuzz(16, 31))
```
