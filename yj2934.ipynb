```python
def getBondPrice(y, face, couponRate, m, ppy=1):
    periods = int(m * ppy)
    coupon = face * couponRate / ppy
    bondPrice = 0.0
    for period in range(1, periods + 1):
        cashflow = coupon
        if period == periods:
            cashflow += face
        bondPrice += cashflow / (1 + y / ppy) ** period

    return bondPrice
```

```python
print(getBondPrice(0.03, 2000000, 0.04, 10, 1))
print(getBondPrice(0.03, 2000000, 0.04, 10, 2))
print(getBondPrice(0.03, 2000000, 0.04, 10))
```

```python
def getBondDuration(y, face, couponRate, m, ppy=1):
    periods = int(m * ppy)
    coupon = face * couponRate / ppy
    bondPrice = 0.0
    weightedPV = 0.0
    for period in range(1, periods + 1):
        cashflow = coupon
        if period == periods:
            cashflow += face
        presentValue = cashflow / (1 + y / ppy) ** period
        bondPrice += presentValue
        weightedPV += (period / ppy) * presentValue

    bondDuration = weightedPV / bondPrice
    return bondDuration
```

```python
print(getBondDuration(0.03, 2000000, 0.04, 10, 1))
print(getBondDuration(0.03, 2000000, 0.04, 10))
print(getBondDuration(0.03, 2000000, 0.04, 10, 2))
```

```python
def getBondPrice_E(face, couponRate, m, yc):
    if len(yc) != m:
        raise ValueError("yc must contain one rate for each year.")

    coupon = face * couponRate
    bondPrice = 0.0
    for index, rate in enumerate(yc):
        time = index + 1
        cashflow = coupon
        if time == m:
            cashflow += face
        bondPrice += cashflow / (1 + rate) ** time

    return bondPrice
```

```python
yc = [0.010, 0.015, 0.020, 0.025, 0.030]
print(getBondPrice_E(2000000, 0.04, 5, yc))
```

```python
def getBondPrice_Z(face, couponRate, times, yc):
    if len(times) != len(yc) or len(times) == 0:
        raise ValueError("times and yc must have the same nonzero length.")

    coupon = face * couponRate
    maturity = times[-1]
    bondPrice = 0.0
    for time, rate in zip(times, yc):
        cashflow = coupon
        if time == maturity:
            cashflow += face
        bondPrice += cashflow / (1 + rate) ** time

    return bondPrice
```

```python
times = [1, 1.5, 3, 4, 7]
yc = [0.010, 0.015, 0.020, 0.025, 0.030]
print(getBondPrice_Z(2000000, 0.04, times, yc))
```

```python
def FizzBuzz(start, finish):
    # Both start and finish are included.
    outlist = []

    for number in range(start, finish + 1):
        if number % 15 == 0:
            outlist.append("fizzbuzz")
        elif number % 3 == 0:
            outlist.append("fizz")
        elif number % 5 == 0:
            outlist.append("buzz")
        else:
            outlist.append(number)

    return outlist
```

```python
print(FizzBuzz(1, 15))
print(FizzBuzz(14, 16))
print(FizzBuzz(15, 15))
```
