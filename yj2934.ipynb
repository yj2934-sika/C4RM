## Question 1: Map

Use map() to return the maximum value of each inner list.

```python
def max_val(input_list):
    """Return the maximum value of each inner list using map()."""
    return list(map(max, input_list))


L = [[1, 2, 5], [4, 3, 3], [1, 7, 6]]

print(max_val(L))
```

## Question 2: Filter

Use filter() to keep only inner lists whose length is exactly 3.

```python
def filter_list(input_list):
    """Return inner lists of length 3 using filter()."""
    return list(
        filter(lambda inner_list: len(inner_list) == 3, input_list)
    )


L = [[2, 5, 4, 9], [2, 2], [3, 4, 5]]

print(filter_list(L))
```

## Question 3: Lambda

Use a lambda function to keep the first inner list unchanged,
square the second inner list, and cube the third inner list.

```python
def expo_list(input_list):
    """Raise values in successive inner lists to powers 1, 2, and 3."""
    return [
        list(map(lambda x: x ** (index + 1), inner_list))
        for index, inner_list in enumerate(input_list)
    ]


L = [[1, 2], [2, 3, 4], [5, 6, 7]]

print(expo_list(L))
```

## Question 4: Comprehension

Use list comprehension to assign:
- 'a' to numbers greater than 8.
- 'b' to numbers greater than 5 and less than or equal to 8.
- 'c' to all other numbers.

```python
def list_comp(input_list):
    """Classify integers using list comprehension."""
    return [
        'a' if x > 8 else 'b' if x > 5 else 'c'
        for x in input_list
    ]


print(list_comp(range(1, 11)))

print("Boundary check:", list_comp([5, 6, 8, 9]))
```

## Question 5: *args and **kwargs

Use the first additional positional argument as the middle name.
Accept optional semester, year, and course keyword arguments.
Return a formatted registration string.

The name order follows the provided examples:
first name, optional middle name, last name.

```python
def register_student(first_name, last_name, *args, **kwargs):
    """Return a registration string using optional args and kwargs."""

    name_parts = [first_name]

    if args:
        name_parts.append(args[0])

    name_parts.append(last_name)
    full_name = ' '.join(name_parts)

    registration_parts = [
        str(kwargs[key])
        for key in ('semester', 'year', 'course')
        if key in kwargs
        and kwargs[key] is not None
        and str(kwargs[key]) != ''
    ]

    registration = ' '.join(registration_parts)

    return f"{full_name} is registered for {registration}".strip()


print(
    register_student(
        'Jane',
        'Doe',
        semester='Summer',
        year='2023'
    )
)

print(
    register_student(
        'John',
        'Doe',
        'Jack',
        semester='Summer',
        year='2023',
        course='AC4RM'
    )
)

print(register_student('Jane', 'Doe'))

print(
    register_student(
        'John',
        'Doe',
        'Jack',
        'Ignored',
        year=2023
    )
)
```

## Question 6: Decorator

Implement a time_it decorator that accepts any positional and keyword
arguments and returns a formatted execution-time string.

```python
from datetime import datetime
from functools import wraps
import time


def time_it(func):
    """Return a decorator that reports execution time as a string."""

    @wraps(func)
    def wrapper(*args, **kwargs):
        start = datetime.now()

        func(*args, **kwargs)

        elapsed = (datetime.now() - start).total_seconds()

        return (
            f"The function {func.__name__} "
            f"takes {elapsed:.5f} seconds"
        )

    return wrapper


# Preserve the decorator before Question 7 reuses the name time_it.
time_it_decorator = time_it


@time_it
def sample_function(delay, *, message):
    """Pause briefly and return a message."""
    time.sleep(delay)
    return message


result = sample_function(0.01, message="Done")

print(result)
print("Return type:", type(result).__name__)
print("Function name:", sample_function.__name__)
```

## Question 7: Context Manager

Implement a time_it context manager that prints the execution time
of the code inside a with statement.

The decorator from Question 6 remains available as time_it_decorator.

```python
import contextlib
from datetime import datetime
import time


@contextlib.contextmanager
def time_it():
    """Print the execution time of a block of code."""

    start = datetime.now()

    try:
        yield

    finally:
        elapsed = (datetime.now() - start).total_seconds()

        print(f"The function takes {elapsed:.5f} seconds")


def context_example(delay):
    """Run a short task inside the timed context."""
    time.sleep(delay)


with time_it() as timer:
    context_example(0.01)
```
